# 数据探索与可视化

在这个Notebook中，我们将对数据集进行初步探索，了解数据的结构和特征，并进行可视化分析。

```python
# 导入必要的库
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# 设置可视化风格
sns.set(style='whitegrid')
```

```python
# 加载数据
data_path = '../data/processed/your_dataset.csv'  # 请替换为实际数据路径
data = pd.read_csv(data_path)

# 显示数据的前几行
data.head()
```

```python
# 数据基本信息
data.info()
```

```python
# 描述性统计
data.describe()
```

```python
# 可视化数据分布
plt.figure(figsize=(12, 6))
sns.histplot(data['your_column'], bins=30, kde=True)  # 请替换为实际列名
plt.title('Your Column Distribution')
plt.xlabel('Your Column')
plt.ylabel('Frequency')
plt.show()
```

```python
# 可视化特征之间的关系
plt.figure(figsize=(12, 6))
sns.scatterplot(x='feature1', y='feature2', data=data)  # 请替换为实际特征名
plt.title('Feature1 vs Feature2')
plt.xlabel('Feature1')
plt.ylabel('Feature2')
plt.show()
```
