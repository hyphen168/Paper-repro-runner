# Training Debug Notebook

This notebook is designed for debugging the training process of the industrial vision model. It includes sections for loading data, initializing the model, training, and evaluating the model.

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt
from src.industrial_vision.data.loaders import load_data
from src.industrial_vision.models.detector import Detector
from src.industrial_vision.pipelines.train import train_model
from src.industrial_vision.pipelines.evaluate import evaluate_model

# Load the dataset
train_data, val_data = load_data(batch_size=32)

# Initialize the model
model = Detector()

# Train the model
num_epochs = 10
for epoch in range(num_epochs):
    print(f"Epoch {epoch + 1}/{num_epochs}")
    train_model(model, train_data)
    evaluate_model(model, val_data)

# Visualize training results
plt.plot(model.loss_history)
plt.title('Training Loss Over Epochs')
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.show()
```
